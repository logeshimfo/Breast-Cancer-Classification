```python
x = df.drop("target" , axis = 1)
y = df["target"]
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42 , stratify = y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps = [
        ("scaler" , StandardScaler()),
        ("model" , LogisticRegression())
    ]
)
```

```python
logistic_model.fit(x_train , y_train)
y_train_pred = logistic_model.predict(x_train)
y_test_pred = logistic_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score , precision_score , recall_score , f1_score , confusion_matrix , classification_report , roc_auc_score
print("Accuracy : " , accuracy_score(y_test , y_test_pred))
print("Precision : " , precision_score(y_test , y_test_pred))
print("Recall : " , recall_score(y_test , y_test_pred))
print("F1 : " , f1_score(y_test , y_test_pred))
```

```python
y_train_pred = logistic_model.predict(x_train)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
```

```python
print("Confusion Matrix : " , confusion_matrix(y_test , y_test_pred))
```

```python
print("Classification Report : " , classification_report(y_test , y_test_pred , target_names = data.target_names))
```

```python
y_test_proba = logistic_model.predict_proba(x_test)[:, 1]
print("ROC-AUC : " , roc_auc_score(y_test , y_test_pred))
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier

knn_model = Pipeline(
    steps = [
        ("Scaler : " , StandardScaler()),
        ("Model : ", KNeighborsClassifier())
    ]
)

knn_model.fit(x_train , y_train)

y_train_pred = knn_model.predict(x_train)
y_test_pred = knn_model.predict(x_test)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision     :", precision_score(y_test, y_test_pred))
print("Recall        :", recall_score(y_test, y_test_pred))
print("F1            :", f1_score(y_test, y_test_pred))

y_test_prob = knn_model.predict_proba(x_test)[:, 1]

print("ROC-AUC       :", roc_auc_score(y_test, y_test_prob))
```

```python
#SVM
from sklearn.svm import SVC

svc_model = Pipeline(
    steps = [
        ("Scaler" , StandardScaler()),
        ("model", SVC(probability = True))
    ]
)

svc_model.fit(x_train , y_train)

y_train_pred = svc_model.predict(x_train)
y_test_pred = svc_model.predict(x_test)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision     :", precision_score(y_test, y_test_pred))
print("Recall        :", recall_score(y_test, y_test_pred))
print("F1            :", f1_score(y_test, y_test_pred))

y_test_prob = svc_model.predict_proba(x_test)[:, 1]

print("ROC-AUC       :", roc_auc_score(y_test, y_test_prob))
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state = 42)

dt_model.fit(x_train , y_train)

y_train_pred = dt_model.predict(x_train)
y_test_pred = dt_model.predict(x_test)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision     :", precision_score(y_test, y_test_pred))
print("Recall        :", recall_score(y_test, y_test_pred))
print("F1            :", f1_score(y_test, y_test_pred))

y_test_prob = dt_model.predict_proba(x_test)[:, 1]

print("ROC-AUC       :", roc_auc_score(y_test, y_test_prob))  

#Dt is overfitting

# Train Accuracy: 1.0
# Test Accuracy : 0.9122807017543859
# Precision     : 0.9558823529411765
# Recall        : 0.9027777777777778
# F1            : 0.9285714285714286
# ROC-AUC       : 0.9156746031746031
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42)

rf_model.fit(x_train, y_train)

y_train_pred = rf_model.predict(x_train)
y_test_pred = rf_model.predict(x_test)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision     :", precision_score(y_test, y_test_pred))
print("Recall        :", recall_score(y_test, y_test_pred))
print("F1            :", f1_score(y_test, y_test_pred))

y_test_prob = rf_model.predict_proba(x_test)[:, 1]

print("ROC-AUC       :", roc_auc_score(y_test, y_test_prob))

#RF is overfitting

# Train Accuracy: 1.0
# Test Accuracy : 0.956140350877193
# Precision     : 0.958904109589041
# Recall        : 0.9722222222222222
# F1            : 0.9655172413793104
# ROC-AUC       : 0.9937169312169312
```

```python
# Gradient Boosting
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(random_state=42)

gb_model.fit(x_train, y_train)

y_train_pred = gb_model.predict(x_train)
y_test_pred = gb_model.predict(x_test)

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision     :", precision_score(y_test, y_test_pred))
print("Recall        :", recall_score(y_test, y_test_pred))
print("F1            :", f1_score(y_test, y_test_pred))

y_test_prob = gb_model.predict_proba(x_test)[:, 1]

print("ROC-AUC       :", roc_auc_score(y_test, y_test_prob))

# Train Accuracy: 1.0
# Test Accuracy : 0.956140350877193
# Precision     : 0.9466666666666667
# Recall        : 0.9861111111111112
# F1            : 0.9659863945578231
# ROC-AUC       : 0.9907407407407408
```

```python
from sklearn.model_selection import cross_val_score
def cross_validation(model , x_train , y_train):
    scores = cross_val_score(
        model,
        x_train,
        y_train,
        cv = 5,
        scoring = "f1"
    )
    print(scores)
    print("Mean F1 : " , scores.mean())
```

```python
cross_validation(logistic_model , x_train , y_train)
cross_validation(knn_model , x_train , y_train)
cross_validation(svc_model , x_train , y_train)
cross_validation(dt_model , x_train , y_train)
cross_validation(rf_model , x_train , y_train)
cross_validation(gb_model , x_train , y_train)
```

```python
#Tune Logistic Regression
from sklearn.model_selection import GridSearchCV
#Parameter Grid
logistic_parm_grid = {
    "model__C" : [0.01 , 0.1 , 1 , 10 , 100], #Regularization small c -> less regularization large(simple model) c -> strong regularization(complex model)
    "model__solver" : ["liblinear" , "lbfgs"]
}

#GridSearchCV
logistic_grid = GridSearchCV(
    estimator = logistic_model, #Our Logistic Regression model
    param_grid = logistic_parm_grid, #Our Parameter Grid
    cv = 5, #5-fold cross-validation
    scoring = "f1", #Use F1-score to decide which hyperparameters are best
    n_jobs = -1 #This controls how many CPU cores/processes can be used
    
)
```

```python
logistic_grid.fit(x_train , y_train)
print("Best Parameters:", logistic_grid.best_params_) #Find the best parameters
print("Best CV F1:", logistic_grid.best_score_) #Find the best CV F1
best_logistic = logistic_grid.best_estimator_ #Take the best Logistic Regression pipeline found by GridSearchCV and store it in best_logistic
```

```python
#Tune Support Vector Classifier
svc_param_grid = {
    "model__C": [0.1, 1, 10, 100],
    "model__kernel": ["linear", "rbf"],
    "model__gamma": ["scale", "auto"]
}

svc_grid = GridSearchCV(
    estimator=svc_model,
    param_grid=svc_param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)
```

```python
svc_grid.fit(x_train, y_train)
print("Best Parameters:", svc_grid.best_params_)
print("Best CV F1:", svc_grid.best_score_)
best_svc = svc_grid.best_estimator_
```

```python
y_test_pred = best_logistic.predict(x_test)

print("Logistic Regression")
print("Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall   :", recall_score(y_test, y_test_pred))
print("F1       :", f1_score(y_test, y_test_pred))

y_test_prob = best_logistic.predict_proba(x_test)[:, 1]

print("ROC-AUC  :", roc_auc_score(y_test, y_test_prob))
```

```python
y_test_pred = best_svc.predict(x_test)

print("SVC")
print("Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall   :", recall_score(y_test, y_test_pred))
print("F1       :", f1_score(y_test, y_test_pred))

y_test_prob = best_svc.predict_proba(x_test)[:, 1]

print("ROC-AUC  :", roc_auc_score(y_test, y_test_prob))
```
